```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from sklearn import datasets

sns.set_theme(style='whitegrid')
print(f'Scikit-Learn Version: {sklearn.__version__}')

# Load Breast Cancer Wisconsin Dataset (Binary Classification)
cancer = datasets.load_breast_cancer()
X, y = cancer.data, cancer.target
print('Breast Cancer Dataset Shape:', X.shape, '| Target Classes:', cancer.target_names)

# Load Wine Dataset (Multi-Class Classification)
wine = datasets.load_wine()
X_wine, y_wine = wine.data, wine.target
print('Wine Dataset Shape         :', X_wine.shape, '| Target Classes:', wine.target_names)
```

## 1. Train/Test Split & Feature Scaling

Splitting the data into training (75%) and testing (25%) sets using `stratify=y` to preserve target distribution, followed by `StandardScaler`.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

print(f'Train set: {X_train.shape} | Test set: {X_test.shape}')
print('Train class balance:', np.bincount(y_train))
print('Test class balance :', np.bincount(y_test))
```

## 2. K-Nearest Neighbors (KNN)

KNN is a non-parametric, instance-based learning algorithm that assigns labels based on the majority vote of nearest neighbours.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_s, y_train)
knn_preds = knn.predict(X_test_s)
print(f'KNN (k=5) Test Accuracy: {accuracy_score(y_test, knn_preds):.4f}')
```

## 3. Logistic Regression

Logistic Regression models the probability of binary outcomes using the sigmoid logistic function.

```python
from sklearn.linear_model import LogisticRegression

logreg = LogisticRegression(max_iter=5000, random_state=42)
logreg.fit(X_train_s, y_train)
logreg_preds = logreg.predict(X_test_s)
print(f'Logistic Regression Test Accuracy: {accuracy_score(y_test, logreg_preds):.4f}')
```

## 4. Decision Trees & Tree Visualization

Decision Trees split features hierarchically based on impurity measures such as Gini Impurity or Information Gain.

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree

dt = DecisionTreeClassifier(max_depth=4, random_state=42)
dt.fit(X_train, y_train)
dt_preds = dt.predict(X_test)
print(f'Decision Tree (max_depth=4) Test Accuracy: {accuracy_score(y_test, dt_preds):.4f}')

plt.figure(figsize=(14, 7))
plot_tree(dt, feature_names=cancer.feature_names, class_names=cancer.target_names, filled=True, rounded=True)
plt.title('Decision Tree Visualization (max_depth=4)')
plt.show()
```

## 5. Random Forest Classifier

Random Forest is an ensemble of decision trees trained on bootstrap samples with random feature selection.

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
rf_preds = rf.predict(X_test)
print(f'Random Forest (n_estimators=100) Test Accuracy: {accuracy_score(y_test, rf_preds):.4f}')
```

## 6. Support Vector Machines (SVM)

SVM finds optimal hyperplanes maximizing margin between classes using kernel tricks (Linear, RBF, Polynomial).

```python
from sklearn.svm import SVC

svm_linear = SVC(kernel='linear', probability=True, random_state=42)
svm_linear.fit(X_train_s, y_train)
print(f'Linear SVM Test Accuracy: {accuracy_score(y_test, svm_linear.predict(X_test_s)):.4f}')

svm_rbf = SVC(kernel='rbf', probability=True, random_state=42)
svm_rbf.fit(X_train_s, y_train)
print(f'RBF Kernel SVM Test Accuracy: {accuracy_score(y_test, svm_rbf.predict(X_test_s)):.4f}')
```

## 7. Naive Bayes Classifiers

Naive Bayes applies Bayes' theorem assuming conditional feature independence.

```python
from sklearn.naive_bayes import GaussianNB, BernoulliNB

gnb = GaussianNB()
gnb.fit(X_train_s, y_train)
print(f'Gaussian Naive Bayes Test Accuracy: {accuracy_score(y_test, gnb.predict(X_test_s)):.4f}')
```

## 8. Boosting Algorithms (Gradient Boosting, HistGradientBoosting, AdaBoost)

Boosting sequentially builds weak learners to minimize loss functions.

```python
from sklearn.ensemble import GradientBoostingClassifier, HistGradientBoostingClassifier, AdaBoostClassifier

gb = GradientBoostingClassifier(n_estimators=100, random_state=42)
gb.fit(X_train, y_train)
print(f'Gradient Boosting Test Accuracy: {accuracy_score(y_test, gb.predict(X_test)):.4f}')

hgb = HistGradientBoostingClassifier(random_state=42)
hgb.fit(X_train, y_train)
print(f'HistGradientBoosting Test Accuracy: {accuracy_score(y_test, hgb.predict(X_test)):.4f}')

ada = AdaBoostClassifier(n_estimators=100, random_state=42)
ada.fit(X_train, y_train)
print(f'AdaBoost Test Accuracy: {accuracy_score(y_test, ada.predict(X_test)):.4f}')
```

## 9. Multi-Class Classification Strategies (OvR vs OvO)

Evaluating One-vs-Rest (OvR) and One-vs-One (OvO) reduction strategies on the Wine multi-class dataset.

```python
from sklearn.multiclass import OneVsRestClassifier, OneVsOneClassifier

X_w_tr, X_w_te, y_w_tr, y_w_te = train_test_split(X_wine, y_wine, test_size=0.3, random_state=42, stratify=y_wine)
scaler_w = StandardScaler()
X_w_tr_s = scaler_w.fit_transform(X_w_tr)
X_w_te_s = scaler_w.transform(X_w_te)

ovr = OneVsRestClassifier(LogisticRegression(max_iter=1000))
ovr.fit(X_w_tr_s, y_w_tr)
print(f'One-vs-Rest (OvR) Test Accuracy: {accuracy_score(y_w_te, ovr.predict(X_w_te_s)):.4f}')

ovo = OneVsOneClassifier(LogisticRegression(max_iter=1000))
ovo.fit(X_w_tr_s, y_w_tr)
print(f'One-vs-One (OvO) Test Accuracy : {accuracy_score(y_w_te, ovo.predict(X_w_te_s)):.4f}')
```

## 10. Classification Performance Metrics & Classification Report

Detailed assessment of model outputs using Accuracy, Precision, Recall, F1-Score, and Classification Report.

```python
from sklearn.metrics import precision_score, recall_score, f1_score, classification_report

print('--- Random Forest Performance ---')
print('Accuracy :', accuracy_score(y_test, rf_preds).round(4))
print('Precision:', precision_score(y_test, rf_preds).round(4))
print('Recall   :', recall_score(y_test, rf_preds).round(4))
print('F1-Score :', f1_score(y_test, rf_preds).round(4))

print('\n--- Classification Report ---\n', classification_report(y_test, rf_preds, target_names=cancer.target_names))
```

## 11. Confusion Matrix Visualization

Visualizing true positives, true negatives, false positives, and false negatives using `ConfusionMatrixDisplay`.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, rf_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=cancer.target_names)
fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(cmap='Blues', ax=ax)
plt.title('Random Forest — Confusion Matrix')
plt.show()
```

## 12. ROC Curve & ROC-AUC Score

The Receiver Operating Characteristic (ROC) curve evaluates True Positive Rate vs False Positive Rate across thresholds.

```python
from sklearn.metrics import roc_curve, roc_auc_score, RocCurveDisplay

y_probs_rf = rf.predict_proba(X_test)[:, 1]
fpr, tpr, thresholds = roc_curve(y_test, y_probs_rf)
auc_score = roc_auc_score(y_test, y_probs_rf)

fig, ax = plt.subplots(figsize=(7, 5))
RocCurveDisplay(fpr=fpr, tpr=tpr, roc_auc=auc_score, estimator_name='Random Forest').plot(ax=ax)
plt.plot([0, 1], [0, 1], 'k--', label='Random Chance (AUC = 0.50)')
plt.title('ROC Curve — Random Forest Classifier')
plt.legend()
plt.show()
```

## 13. Precision-Recall Curve & Average Precision

Precision-Recall (PR) curves are preferred over ROC curves when dealing with highly imbalanced class distributions.

```python
from sklearn.metrics import precision_recall_curve, average_precision_score, PrecisionRecallDisplay

precision, recall, pr_thresholds = precision_recall_curve(y_test, y_probs_rf)
ap_score = average_precision_score(y_test, y_probs_rf)

fig, ax = plt.subplots(figsize=(7, 5))
PrecisionRecallDisplay(precision=precision, recall=recall, average_precision=ap_score, estimator_name='Random Forest').plot(ax=ax)
plt.title('Precision-Recall Curve (AP = {:.4f})'.format(ap_score))
plt.show()
```

## 14. Handling Class Imbalance & Threshold Tuning

Addressing imbalanced target labels using cost-sensitive learning (`class_weight='balanced'`) and decision threshold adjustments.

```python
# Create artificial imbalanced dataset (90% negative, 10% positive)
np.random.seed(42)
imb_idx = np.where(y == 0)[0][:30]
pos_idx = np.where(y == 1)[0]
keep_idx = np.concatenate([imb_idx, pos_idx])
X_imb, y_imb = X[keep_idx], y[keep_idx]

X_imb_tr, X_imb_te, y_imb_tr, y_imb_te = train_test_split(X_imb, y_imb, test_size=0.3, random_state=42, stratify=y_imb)

# Standard Logistic Regression vs Class-Weighted Logistic Regression
lr_std = LogisticRegression(max_iter=1000).fit(X_imb_tr, y_imb_tr)
lr_bal = LogisticRegression(max_iter=1000, class_weight='balanced').fit(X_imb_tr, y_imb_tr)

print('Standard Model Recall     :', recall_score(y_imb_te, lr_std.predict(X_imb_te)).round(4))
print('Balanced Weight Model Recall:', recall_score(y_imb_te, lr_bal.predict(X_imb_te)).round(4))

# Custom Decision Threshold Tuning
probs = lr_bal.predict_proba(X_imb_te)[:, 1]
for thresh in [0.3, 0.5, 0.7]:
    custom_preds = (probs >= thresh).astype(int)
    print(f'Threshold {thresh} -> Precision: {precision_score(y_imb_te, custom_preds):.3f} | Recall: {recall_score(y_imb_te, custom_preds):.3f}')
```

## 15. Probability Calibration & Reliability Curves

Some classifiers (like Support Vector Machines or Naive Bayes) produce uncalibrated probabilities. `CalibratedClassifierCV` maps outputs to true probabilities.

```python
from sklearn.calibration import CalibratedClassifierCV, calibration_curve

# Uncalibrated SVM probabilities
svm_uncal = SVC(kernel='rbf', C=1.0).fit(X_train_s, y_train)
calibrated_svm = CalibratedClassifierCV(svm_uncal, method='sigmoid', cv=3)
calibrated_svm.fit(X_train_s, y_train)

prob_uncal = GaussianNB().fit(X_train_s, y_train).predict_proba(X_test_s)[:, 1]
prob_cal = calibrated_svm.predict_proba(X_test_s)[:, 1]

fraction_of_pos_uncal, mean_predicted_val_uncal = calibration_curve(y_test, prob_uncal, n_bins=10)
fraction_of_pos_cal, mean_predicted_val_cal = calibration_curve(y_test, prob_cal, n_bins=10)

plt.figure(figsize=(7, 5))
plt.plot(mean_predicted_val_uncal, fraction_of_pos_uncal, 's-', label='GaussianNB (Uncalibrated)')
plt.plot(mean_predicted_val_cal, fraction_of_pos_cal, 's-', label='Calibrated SVM (Sigmoid)')
plt.plot([0, 1], [0, 1], 'k--', label='Perfect Calibration')
plt.xlabel('Mean Predicted Probability')
plt.ylabel('Fraction of Positives')
plt.title('Reliability Diagram / Probability Calibration')
plt.legend()
plt.show()
```

## 16. Cross-Validation Strategies (`StratifiedKFold` & `cross_validate`)

Evaluating generalization performance robustly across multiple data splits using cross-validation.

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scoring_metrics = ['accuracy', 'precision', 'recall', 'f1', 'roc_auc']

cv_results = cross_validate(rf, X, y, cv=skf, scoring=scoring_metrics)

print('--- 5-Fold Stratified Cross-Validation Results ---')
for metric in scoring_metrics:
    scores = cv_results[f'test_{metric}']
    print(f'{metric.capitalize():12s}: {scores.mean():.4f} +/- {scores.std():.4f}')
```

## 17. Hyperparameter Optimization (`GridSearchCV` & `RandomizedSearchCV`)

Finding optimal hyperparameter configurations systematically.

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV

param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [3, 5, 8, None],
    'min_samples_split': [2, 5]
}

grid_search = GridSearchCV(RandomForestClassifier(random_state=42), param_grid, cv=3, scoring='f1')
grid_search.fit(X_train, y_train)

print('Best Grid Search Parameters:', grid_search.best_params_)
print(f'Best Grid Search F1-Score : {grid_search.best_score_:.4f}')
```

## 18. Advanced Ensemble Methods (`VotingClassifier` & `StackingClassifier`)

Combining multiple diverse classifiers via Voting or Meta-Learner Stacking.

```python
from sklearn.ensemble import VotingClassifier, StackingClassifier

# 1. Soft Voting Classifier
voting_clf = VotingClassifier(estimators=[
    ('lr', LogisticRegression(max_iter=1000)),
    ('rf', RandomForestClassifier(n_estimators=50, random_state=42)),
    ('gnb', GaussianNB())
], voting='soft')
voting_clf.fit(X_train_s, y_train)
print(f'Soft Voting Test Accuracy : {accuracy_score(y_test, voting_clf.predict(X_test_s)):.4f}')

# 2. Stacking Classifier with Logistic Regression Meta-Learner
stacking_clf = StackingClassifier(estimators=[
    ('rf', RandomForestClassifier(n_estimators=50, random_state=42)),
    ('gb', GradientBoostingClassifier(n_estimators=50, random_state=42)),
    ('svm', SVC(probability=True, random_state=42))
], final_estimator=LogisticRegression())
stacking_clf.fit(X_train_s, y_train)
print(f'Stacking Classifier Test Accuracy: {accuracy_score(y_test, stacking_clf.predict(X_test_s)):.4f}')
```

## 19. Model Interpretability: Feature Importances & Permutation Importance

Understanding which features contribute most to classifier predictions.

```python
from sklearn.inspection import permutation_importance

# Feature Importance from Random Forest
importances = rf.feature_importances_
top_indices = np.argsort(importances)[::-1][:10]

plt.figure(figsize=(9, 4))
sns.barplot(x=importances[top_indices], y=cancer.feature_names[top_indices], palette='viridis')
plt.title('Top 10 Feature Importances (Random Forest)')
plt.xlabel('Gini Importance')
plt.show()

# Permutation Importance
perm_imp = permutation_importance(rf, X_test, y_test, n_repeats=10, random_state=42)
perm_top = np.argsort(perm_imp.importances_mean)[::-1][:5]
print('Top 5 Features by Permutation Importance:')
for idx in perm_top:
    print(f'{cancer.feature_names[idx]:25s}: {perm_imp.importances_mean[idx]:.4f}')
```

## 20. Classification Loss Functions (Log Loss & Brier Score)

Quantifying prediction uncertainty and error using Log Loss and Brier Score.

```python
from sklearn.metrics import log_loss, brier_score_loss

ll_rf = log_loss(y_test, y_probs_rf)
bs_rf = brier_score_loss(y_test, y_probs_rf)

print('--- Loss Function Evaluation (Random Forest) ---')
print(f'Log Loss (Cross-Entropy) : {ll_rf:.4f}')
print(f'Brier Score Loss         : {bs_rf:.4f}')
```

## 21. Decision Boundary Visualization

Visualizing classification decision surfaces in 2D space using PCA components.

```python
from sklearn.decomposition import PCA
from sklearn.inspection import DecisionBoundaryDisplay

# Reduce features to 2D for visualization
pca = PCA(n_components=2, random_state=42)
X_pca_tr = pca.fit_transform(X_train_s)
X_pca_te = pca.transform(X_test_s)

clf_2d = LogisticRegression().fit(X_pca_tr, y_train)

fig, ax = plt.subplots(figsize=(7, 5))
DecisionBoundaryDisplay.from_estimator(clf_2d, X_pca_tr, response_method='predict', cmap='coolwarm', alpha=0.6, ax=ax)
scatter = ax.scatter(X_pca_tr[:, 0], X_pca_tr[:, 1], c=y_train, cmap='coolwarm', edgecolors='k', s=30)
plt.title('2D Logistic Regression Decision Boundary (PCA Features)')
plt.xlabel('PCA Feature 1')
plt.ylabel('PCA Feature 2')
plt.show()
```

## 22. Comprehensive Classifier Benchmark & Summary

Comparing performance metrics across all implemented classification models.

```python
models = {
    'K-Nearest Neighbors': (knn, X_test_s),
    'Logistic Regression': (logreg, X_test_s),
    'Decision Tree': (dt, X_test),
    'Random Forest': (rf, X_test),
    'Linear SVM': (svm_linear, X_test_s),
    'RBF SVM': (svm_rbf, X_test_s),
    'Gaussian Naive Bayes': (gnb, X_test_s),
    'Gradient Boosting': (gb, X_test),
    'HistGradientBoosting': (hgb, X_test),
    'AdaBoost': (ada, X_test),
    'Soft Voting': (voting_clf, X_test_s),
    'Stacking Classifier': (stacking_clf, X_test_s)
}

results = []
for name, (model, X_t) in models.items():
    preds = model.predict(X_t)
    acc = accuracy_score(y_test, preds)
    prec = precision_score(y_test, preds)
    rec = recall_score(y_test, preds)
    f1 = f1_score(y_test, preds)
    results.append({'Model': name, 'Accuracy': acc, 'Precision': prec, 'Recall': rec, 'F1-Score': f1})

benchmark_df = pd.DataFrame(results).sort_values(by='F1-Score', ascending=False)
print(benchmark_df.to_string(index=False))
```

## 23. Scikit-Learn Classification Best Practices Cheat Sheet

| Task | Recommended Transformer / Estimator | Key Parameter / Best Practice |
|---|---|---|
| **Linearly Separable Data** | `LogisticRegression`, `SVC(kernel='linear')` | Set `max_iter>=1000` and tune `C`. |
| **Complex Non-Linear Patterns** | `RandomForestClassifier`, `HistGradientBoostingClassifier` | Use GBDT for high accuracy; tune `learning_rate` & `max_iter`. |
| **Imbalanced Classes** | Any classifier with `class_weight='balanced'` | Evaluate with PR-AUC / Average Precision rather than Accuracy. |
| **Uncalibrated Probabilities** | `CalibratedClassifierCV` | Calibrate SVM/Naive Bayes outputs before thresholding. |
| **Model Ensembling** | `VotingClassifier`, `StackingClassifier` | Combine diverse base models (e.g., Tree + Linear + Probabilistic). |
| **Interpretability** | `DecisionTreeClassifier`, `permutation_importance` | Use Permutation Importance over Gini Importance to avoid cardinality bias. |
